```python
#Imports
import pandas as pd
```

```python
#Read and display raw csv file
br = pd.read_csv("BaseballReference.csv")

display(br)
```

```python
#Takes only the columns needed for my analysis
cleaned_br = br[["Rk", "Tm", "W", "L", "RA"]]
cleaned_br = cleaned_br.iloc[:-1]
#Drops index 30 "averages" row
display(cleaned_br)
```

```python
cleaned_br["Rk"] = cleaned_br["Rk"].astype(int)
cleaned_br = cleaned_br
cleaned_br.dtypes
display(cleaned_br)
```

```python
#Rename columns for better readibility
cleaned_br = cleaned_br.rename(columns={
    "Rk":"Rank",
    "Tm":"Team Name",
    "W":"Wins",
    "L":"Losses",
    "RA":"Runs Allowed"
})

display(cleaned_br)
```

```python
#Writes file into csv 
cleaned_br.to_csv("Baseball_Reference_Cleaned.csv")
```
